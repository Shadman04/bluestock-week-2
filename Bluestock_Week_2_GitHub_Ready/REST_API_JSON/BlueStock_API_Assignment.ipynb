```python
import requests
import json
import pandas as pd

# Public API URL
url = "https://jsonplaceholder.typicode.com/posts"

# Send GET request
response = requests.get(url, params={"_limit": 10})

# Check response status
print("Status Code:", response.status_code)

# Convert response into JSON
data = response.json()

print("\nTotal records received:", len(data))
```

```python
import requests
import json
import pandas as pd

url = "https://api.mfapi.in/mf/125497"

response = requests.get(url, timeout=30)
response.raise_for_status()

data = response.json()

print("Fund:", data["meta"]["scheme_name"])
print("Fund House:", data["meta"]["fund_house"])
print("Status:", data["status"])

# Save complete JSON response
with open("indian_mutual_fund.json", "w") as file:
    json.dump(data, file, indent=4)

print("JSON saved successfully!")
```

```python
# Extract NAV records
df = pd.DataFrame(data["data"])

# Add fund details
df["scheme_name"] = data["meta"]["scheme_name"]
df["fund_house"] = data["meta"]["fund_house"]
df["scheme_code"] = data["meta"]["scheme_code"]

# Convert NAV into numeric values
df["nav"] = pd.to_numeric(df["nav"])

# Save CSV
df.to_csv("indian_mutual_fund.csv", index=False)

# Display first 10 records
display(df.head(10))

print("CSV created successfully!")
```

```python
%%writefile indian_mutual_fund.py

import requests
import json
import pandas as pd

url = "https://api.mfapi.in/mf/125497"

response = requests.get(url, timeout=30)
response.raise_for_status()

data = response.json()

# Save raw JSON
with open("indian_mutual_fund.json", "w") as file:
    json.dump(data, file, indent=4)

# Convert NAV data to DataFrame
df = pd.DataFrame(data["data"])

df["nav"] = pd.to_numeric(df["nav"])
df["scheme_name"] = data["meta"]["scheme_name"]
df["fund_house"] = data["meta"]["fund_house"]
df["scheme_code"] = data["meta"]["scheme_code"]

# Export CSV
df.to_csv("indian_mutual_fund.csv", index=False)

print("API extraction completed successfully!")
print("Total NAV records:", len(df))
print(df.head())
```
